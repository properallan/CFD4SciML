## Gauss-Legendre

A quadratura de Gauss-Legendre (GL) como vimos na aula 02 é definida para o intervalo de integração $[-1, \ 1]$ e tem aproximação dada por
\begin{equation}
    I = \int _{-1} ^1 f(x) dx \approx \sum_{i=1}^n w_i f(x_i)
\end{equation}

em que $n$ é o número de pontos (zeros) que serão utilizados para aproximação, fazendo com que a mesma seja exata até um polinômio de grau $2n-1$.

### Pesos e Raízes para GL

Conforme apresentado na literatura [Legendre-Gauss Quadrature (Wolfram)](https://mathworld.wolfram.com/Legendre-GaussQuadrature.html) é possível encontrar as raízes e os pesos da quadratura utilizando polinômios de Legendre.

Os valores de todas as $n$ raízes $x_i$ são todos os $n$-zeros dos polinômios de Legendre de grau n ($P_n(x)$), isto é:
$$P_n(x_i) = 0$$

Já os pesos ($w_i$) podem ser obtidos de diferentes maneiras, algumas são:
$$w_i = \frac{2}{(1-x_i^2)[P'_n(x_i)]^2} = \frac{2(1-x_i^2)}{(n+1)^2[P_{n+1}(x_i)]^2}$$

---
#### **Exemplo prático (n = 2)**

Na **aula 02** calculamos via integração e resolução de sistema simbólico que os valores para raízes e pesos considerando 2 pontos no contexto do GL são

\begin{gather*}
  w_1 = 1 \hspace{1cm} w_2 = 1 \hspace{1cm} x_1 = -\frac{\sqrt{3}}{3} \hspace{1cm} x_2 = \frac{\sqrt{3}}{3}
\end{gather*}

vamos conferir se com a estratégia de usar os polinômios de Legendre os valores coincidem. Para nosso exemplo, como escolhemos avaliar $n=2$, precisaremos do polinômio $P_2(x)$ e de sua derivada $P_2'(x)$.

Da **aula 01** o polinômio Legendre de grau 2 e sua derivada são

$$P_2(x) = \frac{1}{2}(3x^2-1) \hspace{2cm} P_2'(x) = 3x$$

**Cálculo das Raízes**

Com a formulação apresentada anteriormente para encontrar as raízes devemos fazer $P_2(x) = 0$ o que segue:

$$
\begin{align*}
P_2  = \frac{1}{2}(3x^2-1) = 0 & ⇒ 3x^2 - 1  = 0  \\ x^2  = \frac{1}{3} & ⇒ \boxed{x_{1,2} = \pm \frac{\sqrt{3}}{3}}
\end{align*}
$$

**Cálculo dos Pesos**

Para obter os pesos, vamos usar a estratégia
$$w_i = \frac{2}{(1-x_i^2)[P'_n(x_i)]^2} $$
na ocasião temos então (n=2)
$$w_i = \frac{2}{(1-x_i^2)[3x_i]^2} $$
Para **i = 1** temos $x_1 = -\frac{\sqrt{3}}{3}$
$$w_1 = \frac{2}{(1-(-\frac{\sqrt{3}}{3})^2)[3(-\frac{\sqrt{3}}{3})]^2} = \dfrac{2}{(1-\frac{1}{3})[3]} = \dfrac{2}{\frac{2}{3}[3]} = 1 \ ⟹  \ \boxed{w_1 = 1}$$

Para **i = 2** temos $x_2 = \frac{\sqrt{3}}{3}$
$$w_2 = \frac{2}{(1-(\frac{\sqrt{3}}{3})^2)[3(\frac{\sqrt{3}}{3})]^2} = \dfrac{2}{(1-\frac{1}{3})[3]} = \dfrac{2}{\frac{2}{3}[3]} = 1 \ ⟹  \ \boxed{w_2 = 1} $$

exatamente os mesmos valores obtidos pela estratégia de sistemas e integrais!!

## Gauss Lobatto
Vimos também na **aula 02** que uma quadratura do tipo Gauss-Lobatto (GLL) é aquela que considera em sua aproximação o cálculo da função a ser integrada nas fronteiras, podendo ser escrita da seguinte maneira
\begin{gather*}
    I = \int _{-1} ^1 f(x) dx \approx w_1 f(-1) + \sum_{i = 2} ^{n-1} w_i f(x_i) + w_n f(1)  
\end{gather*}

A quantidade de pontos $n$ ao qual se deseja usar na quadratura GLL, garante exatidão para polinômios até grau $2n-3$.

> **Note:** Se o $n$ escolhido for igual a 2, só serão consideradas as fronteiras!

### Pesos e Raizes para GLL
Conforme apresentado na literatura [Quadratura Gauss Lobatto](https://mathworld.wolfram.com/LobattoQuadrature.html) as raízes internas $(x_i)$ são, por definição matemática, as raízes da *derivada* do **Polinômio de Legendre** de grau $n-1$, ou seja:
$$P'_{n-1}(x_i) = 0$$

Os **pesos** correspondentes também possuem uma formulação própria e ligeiramente diferente a do GL:
$$w_i = \frac{2}{n(n-1) [P_{n-1}(x_i)]^2}$$

No caso dos pesos para os extremos é expressa por:
$$w_{1,n} = \frac{2}{n(n-1)}$$

---
#### **Exemplo Prático ($n=3$ pontos)**
Vamos construir a GLL para 3 pontos. Sabemos que da **aula 02** os valores a serem encontrados são

\begin{gather*}
  w_1 = \frac{1}{3} \hspace{1cm} w_2 = \frac{4}{3} \hspace{1cm} w_3 = \frac{1}{3} \hspace{1cm} x_2 = 0
\end{gather*}

vamos usar a estratégia com polinômios de Legendre.

**Raízes**

Sabemos que duas raízes estão travadas: $x_1 = -1$ e $x_3 = 1$, então nos resta o ponto interno, da literatura ele deve ser a raiz de
$$P'_{\mathbf{3}-1}(x) = P'_{2}(x) = 0$$
como $P_2(x) = \frac{1}{2}(3x^2 - 1)$ então a sua derivada é $P'_2(x) = 3x$.

Igualando a zero, a raiz interna é

$$3x = 0  \Longrightarrow \boxed{x_2 = 0}$$


**Pesos**

Os pesos nos extremos necessitam somente da quantidade de pontos que estamos utilizando, no nosso exemplo, n = 3 assim:
$$w_{1,3} = \frac{2}{3(3-1)} = \frac{1}{3} \ \Longrightarrow \boxed{w_{1,3} = \frac{1}{3}}$$

Já para o peso interno utilizamos o resultado da raiz encontrada ($x_2 = 0$) aplicada no polinômio de grau anterior a $n=3$

$$w_2 = \frac{2}{3(3-1) [P_{2}(x_2)]^2} = \frac{1}{3[\frac{1}{2}(3\cdot 0^2 -1)]^2} = \frac{4}{3} \Longrightarrow \boxed{w_{2} = \frac{4}{3}}$$

novamente os resultados coincidem!!

---

> Fica claro que utilizar os polinômios de Legendre parece a melhor estratégia, sem a necessidade de calcular sistemas não-lineares.

## Generalização das Quadraturas
Existe um adendo apresentado na literatura do *Wolfram* sobre o uso dos polinômios de Legendre:
> "*A Gaussian quadrature with **weighting** function W(x)=1* "

esta frase indica que estamos trabalhando em um contexto no qual a quadratura pode ser escrita como

$$I = \int _{-1} ^1 1 \cdot f(x) dx$$

Se lembrarmos da **aula 01**, o Polinômio de Jacobi está associado a um peso da forma
$$W(x) = (1 - x)^\alpha(1+x)^\beta$$
que para o caso de $\alpha = \beta = 0 → W(x) = 1$ temos a família dos **Polinômios de Legendre** $(P^{0,0}_n (x))$.

Nessa mesma linha, podemos usar a ponderação associada a Jacobi $W(x)$ desde que $\alpha > -1$ e $\beta > -1$ e definir uma família de Quadraturas generalizadas denominadas **Tipo Gauss**.

### **Integração Tipo Gauss**
**Notas**  
> * Para seguir a escrita da literatura usaremos, a partir de agora, $Q$ no lugar de $n$ ao se referir ao número de pontos
> * Nossos sub-indices começam em 0

Uma função $f(x)$ ponderada por $W(x) = (1 - x)^\alpha(1+x)^\beta$ no intervalo $[-1, 1]$ pode ser numericamente calculada como segue

$$\int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta f(x) dx = \sum_{i=0}^{Q-1} w_i ^{\alpha,\beta}f(x_i) + R(f)$$

em que $R(f) = 0$ se $f(x) \in \mathcal{P}_{2Q-k}([-1, 1])$.

Perceba, se:
* $k = 1$ : Temos a quadratura de **Gauss-Jacobi** e se $(\alpha = \beta = 0)$ **Gauss-Legendre**
* $k = 2$ : Temos a quadratura de Gauss-Radau
* $k = 3$ : Temos a quadratura de **Gauss-Lobatto**


Para derivar a expressão anterior nós consideramos o caso em que $f(x)$ é um polinômio de ordem $2Q-k$ (isso é, $f(x) \in \mathcal{P}_{2Q-k}([-1,1])$).

Podemos decompor $f(x)$ em termos de um polinômio interpolador e um erro de interpolação (**veja demonstração ao fim do Notebook**), isto é

$$f(x) = \sum_{i=0}^{Q-1} f(x_i)h_i(x) + s(x)r(x)$$

sendo
$$
\begin{align*}
h_i(x_j) = \delta_{ij}, \ \ h(x) \in & \ \mathcal{P}_{Q-1} \\
s(x_i) = 0, \ \ s(x) \in & \ \mathcal{P}_{Q} \\
r(x) \in & \ \mathcal{P}_{Q-k}
\end{align*}
$$

em que $h_i(x_j)$ é o **polinômio de Lagrange** calculado nos $Q$ pontos nodais $x_i$ com propriedade $\delta_{ij}$, que é a delta de Kronecker, definida como

$$\delta_{ij} = \begin{cases}1, & i = j \\ 0, & i \neq j\end{cases}$$

já $s(x)$ é um polinômio de ordem $Q$ com raízes nos pontos nodais $x_i$, e o polinômio resto $r(x)$ de ordem $Q-k$.

Podemos integrar a equação decomposta com respeito a $(1-x)^\alpha (1+x)^\beta$ no intervalo $[-1, 1]$


$$\int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta f(x)dx = \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta \left (\sum_{i=0}^{Q-1} f(x_i)h_i(x) + s(x)r(x) \right) dx$$

se distribuimos a integral e rearranjamos
$$ \sum_{i=0}^{Q-1} f(x_i) \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta h_i(x)dx + \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta s(x)r(x) dx$$


ao comparar com $\sum_{i=0}^{Q-1} w_i ^{\alpha,\beta}f(x_i) + R(f)$ concluimos que

$$\begin{align*}
w_i ^{\alpha,\beta} =& \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta h_i(x)dx\\
R(f) = & \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta s(x)r(x) dx
\end{align*}
$$

Se $s(x)$ for especificado então os pontos nodais $x_i$ se tornam raízes de $s(x)$ ou seja $s(x_i) = 0$. Especificar $s(x)$ é preferível do que especificar os pontos nodais $x_i$ porque uma escolha apropriada de $s(x)$ torna $R(f) = 0$. Por exemplo, se $s(x) = P^{\alpha, \beta}_Q(x)$, em que $P^{\alpha, \beta}_Q(x)$ é um polinômio de Jacobi, então

$$R(f) =  \int _{-1} ^1 (1 - x)^\alpha(1+x)^\beta P^{\alpha, \beta}_Q(x)r(x) dx$$

Esta integral é a **relação de ortogonalidade** para $P^{\alpha, \beta}_Q(x)$, assim, se $k=1$, então $r(x)$ é um polinômio de ordem $Q-1$ e portanto $R(f) = 0$.

Essa escolha de $s(x)$ determina as clássicas **quadraturas de Gauss** e portanto os pontos nodais são determinados por $x_i = x_{i,P}^{\alpha, \beta}$, em que $x_{i,P}^{\alpha, \beta}$ denotam os $m$ zeros do polinômio de Jacobi $P_{P}^{\alpha, \beta}$ ou seja

$$P_{P}^{\alpha, \beta}(x_{i,P}^{\alpha, \beta}) = 0, \ \ i=0,1, \dots , P-1$$

em que

$$x_{0,P}^{\alpha, \beta}< x_{1,P}^{\alpha, \beta} < \cdots < x_{P-1,P}^{\alpha, \beta}$$

> As integrações de Gauss-Radau e Gauss-Lobatto requerem pontos nodais que incluem um ou ambos os extremos $x=\pm 1$. Essas restrições modificam a forma de $s(x)$ e tem o efeito de alterar a função peso $W(x) = (1 - x)^\alpha(1+x)^\beta$.  
Portanto os pontos nodais são determinados por polinômios de Jacobi de **diferentes valores** de $\alpha$ e $\beta$.

---

* Para a construção do nosso solver DG vamos usar os Polinômios de Jacobi com valores de $k=1$ (GL) e $k=3$ (GLL), para que assim  $R(f) = 0$.



### **Pesos e Raízes das Quadraturas Tipo Gauss**
Assim como o que é apresentado no site do Wolfram, podemos retirar relações mais generalistas a depender da família da quadratura, então conforme Sherwin & Karniadakis (2005) tem-se:

#### **Quadratura tipo Gauss-Jacobi (k=1)**  
Esse tipo de quadratura é a que se enquadra a GL, ou seja, não se considera o calculo da função nos extremos e a integral é exata para funções de grau até $2Q-1$.

---

**Raízes**  
As raízes são exatamente os zeros do polinômio $\mathcal{P}^{\alpha, \beta}_Q(x)$ isto é
$$\boxed{x_i = x_{i,Q}^{\alpha, \beta} \hspace{1cm} \text{tal que} \hspace{1cm}  \mathcal{P}^{\alpha, \beta}_Q(x_{i,Q}^{\alpha, \beta}) = 0 }$$

* Se quisermos integrar um polinômio de ordem 3 com exatidão:  
$$2Q - 1 = 3$$
então $Q = 2$.  
* Se quisermos integrar um polinômio de ordem 4 com exatidão:  
$$2Q - 1 = 4$$
então $Q = 2.5$. Como o valor deve ser inteiro usamos e vimos que para $Q = 2$ a precisão só vale até um polinômio de grau 3, devemos utilizar a **função teto** `ceil` e assim $Q = 3$.  
* A **quantidade de raízes** para quadratura de Gauss-Jacobi é **igual** ao **grau do polinômio de Jacobi** a ser utilizado.

---

**Pesos**  
Assim como apresentado para o caso específico da familia de Legendre, aqui iremos usar a primeira derivada do polinômio. Lembre-se, para o caso de usar polinômios de Legendre:
$$w_i = \frac{2}{(1-x_i^2)[\frac{d}{dx}P^{0,0}_Q(x_i)]^2} $$

No entanto, como estamos considerando a possibilidade de outras famílias (tipo a de Chebyshev) existem outros coeficientes nesta fórmula, o que segue:

$$\boxed{w_{i,Q}^{\alpha, \beta} = \frac{2^{\alpha + \beta + 1} \Gamma(\alpha + Q + 1)\Gamma(\beta + Q + 1)}{Q!\Gamma(\alpha + \beta + Q + 1)(1-x_i^2)[\frac{d}{dx} \left . P^{\alpha,\beta}_Q(x)\right|_{x = x_i}]^2}}$$

* Para encontrar os pesos na quadratura de Gauss-Jacobi usamos **todas as raízes** obtidas para a mesma quadratura
* A raiz $x_i$ ajudará a retornar o peso $w_i$
* O **grau do polinômio de Jacobi** a ser derivado é **igual** ao **número de pontos** para a quadratura ser exata

#### **Quadratura tipo Gauss-Lobatto-Jacobi-Bouzitat de 2ª espécie (k=3)**  
Esse tipo de quadratura é a que se enquadra a GLL, ou seja,  considera o calculo da função nos extremos e, portanto, a solução será exata para polinômios de grau $2Q-3$.

> **Nota:** A de 1ª espécie avalia somente um dos extremos (fica de sugestão a leitura do Apêndice B de Sherwin & Karniadakis)

---
**Raízes**  
Se na versão retirada do Wolfram usamos os zeros referentes aos pontos centrais advindos da derivada de Legendre, para um escopo mais geral podemos utilizar **relações de recorrência** e descobrir que os zeros que queremos encontrar (além dos extremos) estão associados a $P^{\alpha + 1, \beta + 1}_{Q-2}(x)$ (**veja o motivo ao fim do notebook**) ou seja as raízes são:
$$
\boxed{
x_i = \begin{cases}
-1, & i = 0 \\
x^{\alpha + 1, \beta + 1}_{i-1,Q-2}, & i = 1, \dots, Q - 2\\
1,& i = Q-1
\end{cases}}
$$

* Se quisermos integrar um polinômio de **grau 5** então
$$2Q-3 = 5$$
e assim $Q = 4$.
* Se quisermos integrar um polinômio de **grau 6** então
$$2Q-3 = 6$$
e assim $Q = 4.5$. Como o valor deve ser inteiro usamos e vimos que para $Q = 4$ a precisão só vale até um polinômio de grau 5, devemos utilizar a **função teto** `ceil` e assim $Q = 5$.
* Como já temos **2 raízes** (extremos) a quantidade de raízes internas será dada por um **polinômio de Jacobi** de **grau $Q-2$** com coeficientes $\alpha$ e $\beta$ deslocados em $+1$.

---

**Pesos**  
Os pesos seguem uma lógica similar ao caso anterior e são dados por
$$
\boxed{
w_i^{\alpha,\beta} = \begin{cases}
(\beta + 1)C_{0,Q-2}^{\alpha, \beta}, & i = 0 \\
C^{\alpha, \beta}_{i,Q-2}, & i = 1, \dots, Q - 2\\
(\alpha + 1)C_{Q-1,Q-2}^{\alpha, \beta},& i = Q-1
\end{cases}}
$$
no qual se define ainda
$$C^{\alpha, \beta}_{i,Q-2} = \dfrac{2^{\alpha+\beta + 1} \Gamma(\alpha + Q) \Gamma(\beta + Q)}{(Q-1)(Q-1)!\Gamma(\alpha + \beta + Q + 1)[P^{\alpha,\beta}_{Q-1}(x_i)]^2}$$

* Para encontrar os pesos na quadratura de Gauss-Jacobi-Lobatto usamos **todas as raízes** obtidas para a mesma quadratura
* A raiz $x_i$ ajudará a retornar o peso $w_i$
* O **grau do polinômio de Jacobi** a ser utilizado é **igual** a **(Q - 1)** em que $Q$ é o número de pontos para a quadratura ser exata.



## **Estratégia Iterativa e Sequencial**

Poderiamos agora começar a programar já que temos os recursos em mãos para o formato generalista. Mas digo que isso seria um tiro no pé!

Se você está acompanhando com veemência, deve ter se questionado:

> **Como eu encontro TODOS os zeros dos Polinômios de ordens maiores?**

Vimos que para os **pesos** temos **fórmulas fechadas**, mas para as raízes é preciso determinar os zeros nos polinômios.

Para polinômios de grau até 3, é relativamente simples encontrar, podendo ser feito manualmente o processo. Mas nós queremos raízes de polinômios maiores que 5!

Para isso devemos utilizar mais **dois recursos** matemáticos (eu prometo, essas são as últimas teorias dessa aula 😢)

1. **Método de Newton-Raphson:** O método de Newton-Raphson (NR) irá nos ajudar a encontrar as raízes dos Polinômios de forma **Iterativa**. No entanto, somente com ele não garantimos que todas sejam encontradas, justamente por questões de inicialização do método e do formato dos polinômios, por isso é necessário acoplar a Deflação Polinomial.
2. **Deflação Polinomial:** A estratégia de Deflação Polinomial em conjunto com NR vai nos ajudar a encontrar TODAS as raízes SEM REPETIÇÕES de forma **Sequencial**.


### **Deflação Polinomial**

**Objetivo:** Encontrar todas as raízes $x_i = x_{i,m}^{\alpha,\beta}$ tal que $P_m^{\alpha, \beta}(x_{i,m}^{\alpha, \beta}) = 0$. Aqui, por conveniência de escrita, usaremos $m$ sendo o grau do Polinômio de Jacobi.

A estratégia é usada para achar novas raízes a partir de **raízes conhecidas**! No caso devemos fatorar o polinômio inicial com relação a cada raiz conhecida, isso significa que o algoritmo de busca de raízes pode ser aplicado a um formalismo expresso por

$$f_{m-n} = \dfrac{P_m^{\alpha, \beta}(x)}{∏_{i=0}^{n-1}(x-x_i)}$$

---
**Exemplo Ilustrativo**  
Vamos supor que queremos saber todas as raízes do Polinômio $P_2(x) = x^2 - 1$. A partir de um **chute inicial**, o nosso método de NR encontra que uma das **raízes é 1 ($x_0 = 1$)** então agora temos **uma** raíz conhecida, logo n = 1

$$f_{2-1} = \dfrac{x^2-1}{∏_{i=0}^{0}(x-x_i)} = \dfrac{x^2-1}{(x-x_0)} = \dfrac{x^2-1}{(x-1)} = x+1$$

Agora somos capazes de usar esse novo Polinômio no método de NR, o que retornará é claro $x_1 = -1$.

Veja o formato de $f_{m-n}$ vai se alterando (seu grau vai diminuindo) ao decorrer da deflação e iteratividade do NR.  Começamos sem nenhuma raiz conhecida
$$n=0 → f_2(x) = x^2 -1,$$
usamos $f_2$ no método NR e passamos a ter **uma** raiz conhecida
$$n=1 → f_1(x)= x+1$$
usamos novamente o NR para obter a raiz de $f_1$ obtendo assim todas as raízes da função $(x_0 = 1, x_1 = -1)$

---

Antes de modelar essa estratégia junto com o NR vamos relembrar rapidamente como ele funciona.




### **Método de Newton-Raphson**

O método de Newton-Raphson é uma estratégia iterativa para encontrar zeros de funções, ou seja, com ele conseguimos encontrar $x$ tal que $f(x) = 0$, para isso usamos sua definição que é dada por

$$x_{k+1} = x_{k} - \dfrac{f(x_{k})}{f'(x_{k})}$$

sendo que queremos que em algum momento $x_{k+1} - x_{k} < \epsilon$ no qual $\epsilon$ é a tolerância adotada para iteração.
Note que precisamos fornecer para o modelo um valor inicial ($x_0$).

> Como no nosso contexto estamos trabalhando com as familias de Jacobi um **bom chute inicial**, que é definido na literatura de forma explicita, são os zeros de Chebyshev

$$
\boxed{
x_{i,m}^{-1/2,-1/2} = - \cos \left(\dfrac{2i+1}{2m}\pi \right)}
$$


### **Juntando a Deflação com o Método de Newton-Raphson**

Do método NR precisamos calcular a razão entre a função e a sua derivada. Associada a deflação essa razão tem a forma:
$$\dfrac{f_{m-n}(x_{k})}{f'_{m-n}(x_{k})}$$
vamos encontrar uma melhor forma de expressá-la de acordo com os métodos que já desenvolvemos.

---
A equação da deflação pode ser expressa abrindo os termos como:
$$f_{m-n}(x) = \dfrac{P_m^{\alpha, \beta}(x)}{∏_{i=0}^{n-1}(x-x_i)} = \dfrac{P_m^{\alpha, \beta}(x)}{(x-x_0)(x-x_1) \dots (x-x_{n-1})} $$

Derivar isso em um primeiro momento parece um tanto assustador, uma estratégia inteligente, é usar propriedades de funções logaritmicas antes, veja

> para melhor visualização vamos assumir $u(x) = f_{m-n}(x)$ e $p(x) = P_m^{\alpha, \beta}(x)$

$$\ln (u(x))  = \ln \left(\dfrac{p(x)}{(x-x_0)(x-x_1) \dots (x-x_{n-1})} \right) = \ln(p(x)) - [\ln(x-x_0) + \ln(x-x_1) + ⋯ + \ln(x-x_{n-1})]$$

Ao derivar ambos os lados com relação a x temos que:

* A derivada de $\ln(u(x))$ é $\frac{u'(x)}{u(x)}$
* A derivada de $\ln(f(x))$ é $\frac{p'(x)}{p(x)}$
* A derivada de $\ln(x-x_i)$ é $\frac{1}{x-x_i}$

Logo é possível escrever a derivada como:
$$\frac{u'(x)}{u(x)} = \frac{p'(x)}{p(x)} - \sum_{i=0}^{n-1} \frac{1}{x-x_i}$$

ou, voltando a nossa nomenclatura inicial
$$\frac{f_{m-n}'(x)}{f_{m-n}(x)} = \frac{[P_m^{\alpha, \beta}(x)]'}{P_m^{\alpha, \beta}(x)} - \sum_{i=0}^{n-1} \frac{1}{x-x_i}$$

**VEJA SÓ** o lado esquerdo da equação $\frac{f_{m-n}'(x)}{f_{m-n}(x)}$ é exatamente o inverso do que queremos!!

---
Por facilidade, vamos chamar o somatório de $S = \sum_{i=0}^{n-1} \frac{1}{x-x_i}$ e fazer a manipulação
$$\frac{f_{m-n}'(x)}{f_{m-n}(x)} = \frac{[P_m^{\alpha, \beta}(x)]'}{P_m^{\alpha, \beta}(x)} - S = \frac{[P_m^{\alpha, \beta}(x)]' - SP_m^{\alpha, \beta}(x)}{P_m^{\alpha, \beta}(x)} ⟹ \boxed{\frac{f_{m-n}(x)}{f_{m-n}'(x)} = \frac{P_m^{\alpha, \beta}(x)}{[P_m^{\alpha, \beta}(x)]' - SP_m^{\alpha, \beta}(x)}}$$

Por fim, a nossa deflação junto com o método de NR pode ser feita.

$$x_{k+1} = x_{k} - \frac{P_m^{\alpha, \beta}(x_k)}{[P_m^{\alpha, \beta}(x_k)]' - SP_m^{\alpha, \beta}(x_k)}$$

**Ufa, agora podemos juntar todas as peças do quebra-cabeça e começar a codar!!**



## Códigos
### Polinômio de Jacobi e Derivada

Vamos trazer aqui os códigos relacionados aos Polinômios de Jacobi e sua Derivada de 1 ordem.

> Consulte a Aula 01 para mais informações

```python
import numpy as np
import matplotlib.pyplot as plt
import math

### POLINOMIOS DE JACOBI PARA ENCONTRAR DIFERENTES FAMILIAS
# alpha = 0 | beta = 0 --> Polinomios de Legendre
# alpha = -0.5 | beta = -0.5 --> Polinomios de Chebyshev normalizados
def JacobiP(r: float, m: int, alpha: float, beta: float) -> float:
    """
    Avalia o Polinômio de Jacobi de ordem m no ponto r com pesos alpha e beta.

    ### Parameters
      r (float): O ponto a ser avaliado.
      m (int): O grau do polinomio.
      alpha (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      beta (float): Parâmetro associado a alguma família dos polinômios de Jacobi

    ### Returns
      float: Retorna o resultado do polinomio de Jacobi para o ponto r.

    """
    if m == 0:
        return 1.0

    pn0 = 1.0
    pn1 = 0.5 * (alpha - beta + (alpha + beta + 2.0) * r)

    if m == 1:
        return pn1

    pn2 = 0.0
    for n in range(1, m):
        a1n = 2.0 * (n + 1) * (n + alpha + beta + 1) * (2.0 * n + alpha + beta)
        a2n = (2.0 * n + alpha + beta + 1) * (alpha**2 - beta**2)
        a3n = (2.0 * n + alpha + beta) * (2.0 * n + alpha + beta + 1) * (2.0 * n + alpha + beta + 2)
        a4n = 2.0 * (n + alpha) * (n + beta) * (2.0 * n + alpha + beta + 2)

        pn2 = (1.0 / a1n) * ((a2n + a3n * r) * pn1 - a4n * pn0)
        pn0 = pn1
        pn1 = pn2

    return pn2

### DERIVADA DE ORDEM 1 DOS POLINOMIOS DE JACOBI
def DJacobiP(r: float, m: int, alpha: float, beta: float) -> float:
    """
    Primeira derivada do Polinômio de Jacobi de ordem m avaliado no ponto r.
    Utiliza math.gamma para suportar adequadamente os parâmetros fracionários
    (ex: Chebyshev, onde alpha e beta = -0.5).
    """
    if m == 0:
        return 0.0

    if np.isclose(r, -1.0):
        c1 = ((-1.0)**(m-1))*0.5*(alpha+beta+m+1)
        return c1 * (
            math.gamma(m+beta+1) /
            (math.gamma(beta+2)*math.gamma(m))
        )

    if np.isclose(r, 1.0):
        c1 = 0.5*(alpha+beta+m+1)
        return c1 * (
            math.gamma(m+alpha+1) /
            (math.gamma(alpha+2)*math.gamma(m))
        )

    # Para pontos internos, calculamos o polinômio e usamos a relação para a derivada
    Pm  = JacobiP(r, m, alpha, beta)
    Pm1 = JacobiP(r, m-1, alpha, beta)

    b1n = (2*m + alpha + beta)*(1-r**2)
    b2n = m*(alpha-beta-(2*m+alpha+beta)*r)
    b3n = 2*(m+alpha)*(m+beta)

    return (b2n*Pm + b3n*Pm1)/b1n
```

### Raízes para Quadraturas via NR e Delflação

As raízes são obtidas a depender do grau do polinômio

```python
import math
import numpy as np

def jacobi_roots(m: int, alpha: float, beta: float) -> np.ndarray:
    """
    Encontra as raízes do Polinômio de Jacobi de ordem m
    usando Newton-Raphson acoplado com Deflação Polinomial.

    ### Args:
      m (int): O grau do polinomio de Jacobi.
      alpha (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      beta (float): Parâmetro associado a alguma família dos polinômios de Jacobi

    ### Returns:
      array: As m-raízes do Polinomio de Jacobi de grau m
    """
    if m == 0:
        return np.array([])

    x = np.zeros(m)
    epsilon = 1.0e-16

    for k in range(m):
        # Chute inicial: raízes do polinômio de Chebyshev
        r = -np.cos(np.pi * (2.0 * k + 1.0) / (2.0 * m))
        if k > 0:
            r = 0.5 * (r + x[k - 1])

        delta = 1.0
        while abs(delta) > epsilon:
            s = 0.0
            for i in range(k):
                s += 1.0 / (r - x[i])

            pm = JacobiP(r, m, alpha, beta)
            dpm = DJacobiP(r, m, alpha, beta)

            # Relação de Newton-Raphson com Deflação
            delta = -pm / (dpm - pm * s)
            r += delta

        x[k] = r

    return x
```

### Pesos para as Quadraturas

Os pesos estão associados a quantos pontos a quadratura precisa pra ser exata!

```python
def jacobi_gauss_weights(x: np.ndarray, alpha: float, beta: float, n_points: int, quad_type: str = 'GL') -> np.ndarray:
    """
    Calcula os pesos correspondentes para as raízes da Quadratura
    do tipo Gauss-Legendre (GL) ou Gauss-Lobatto-Legendre (GLL) baseando-se no número de pontos.

    ### Args:
      x (np.array): Vetor de pontos a serem avaliados (zeros da quadratura).
      alpha (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      beta (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      n_points (int): Número de pontos para a quadratura ser exata
    ### Returns:
      array: Os pesos da quadratura
    """
    w = np.zeros(n_points)

    if quad_type == 'GL':
        c1 = (2.0**(alpha + beta + 1.0)) * math.gamma(alpha + n_points + 1.0) * math.gamma(beta + n_points + 1.0)
        c2 = math.gamma(n_points + 1.0) * math.gamma(alpha + beta + n_points + 1.0)
        coef = c1 / c2

        for i in range(n_points):
            dpm = DJacobiP(x[i], n_points, alpha, beta)
            w[i] = coef / ((1.0 - x[i]**2) * (dpm**2))

    elif quad_type == 'GLL':
        c1 = (2.0**(alpha + beta + 1.0)) * math.gamma(alpha + n_points) * math.gamma(beta + n_points)
        c2 = (n_points - 1.0)* math.gamma(n_points) * math.gamma(alpha + beta + n_points + 1.0)
        coef = c1 / c2

        for i in range(n_points):
            pm_minus_1 = JacobiP(x[i], n_points - 1, alpha, beta)
            c_i = coef / (pm_minus_1**2)

            if i == 0:
                w[i] = (beta + 1.0) * c_i
            elif i == n_points - 1:
                w[i] = (alpha + 1.0) * c_i
            else:
                w[i] = c_i
    else:
        raise ValueError("Integração inválida. Escolha 'GL' ou 'GLL'.")

    return w
```

### Juntando Raízes e Pesos

**Gauss-Legendre**

A quadratura é exata para um polinômio de grau $P$ em que
$$P = 2n_{GL} -1$$
e $n_{GL}$ são a quantidade de pontos necessários. Logo:
$$n_{GL} = \frac{P + 1}{2}$$

**Gauss-Lobatto**

A quadratura é exata para um polinômio de grau $P$ em que
$$P = 2n_{GLL} -3$$
e $n_{GLL}$ são a quantidade de pontos necessários. Logo:
$$n_{GLL} = \frac{P + 3}{2}$$



```python
def jacobi_gauss_quad(P: int, alpha: float = 0.0, beta: float = 0.0, quad_type: str = 'GL'):
    """
    Retorna as matrizes de raízes (xi) e pesos (wi) garantindo exatidão
    para a integração de um polinômio de grau P.

    Args:
      P (int): O grau do polinomio a ser integrado.
      alpha (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      beta (float): Parâmetro associado a alguma família dos polinômios de Jacobi
      quad_type (str): Tipo de quadratura que deseja utilizar (por padrão utiliza-se Gauss-Legendre (GL))
    """
    if quad_type == 'GL':
        # GL integra exatamente até grau 2n - 1
        # P <= 2n - 1  -->  n >= (P + 1) / 2
        n_points = max(1, math.ceil((P + 1.0) / 2.0))
        # Para xi_GL, o grau do polinomio que da as raizes coincide com o numero de pontos
        xi = jacobi_roots(n_points, alpha, beta)
        wi = jacobi_gauss_weights(xi, alpha, beta, n_points, quad_type='GL')

    elif quad_type == 'GLL':
        # GLL integra exatamente até grau 2n - 3
        # P <= 2n - 3  -->  n >= (P + 3) / 2
        n_points = max(2, math.ceil((P + 3.0) / 2.0))

        # Quantidade de pontos para a quadratura GLL e definicao dos extremos
        xi = np.zeros(n_points)
        xi[0] = -1.0
        xi[-1] = 1.0

        if n_points > 2:
            # Para xi_GLL internos, o grau do polinomio que da as raizes se torna
            # n_pontos - 2, visto que os extremos sao pre-definidos
            x_internos = jacobi_roots(n_points - 2, alpha + 1.0, beta + 1.0)
            xi[1:-1] = x_internos

        wi = jacobi_gauss_weights(xi, alpha, beta, n_points, quad_type='GLL')

    else:
        raise ValueError("Integração inválida. Escolha 'GL' ou 'GLL'.")

    return xi, wi
```

## Exemplos de Cálculo de Quadratura

Vamos usar nosso código para fazer o cálculo de integrais para função com ordens maiores do que 5!

Antes disso, vamos criar um script que seja capaz de receber:
* A função a ser integrada $f(x)$
* O intervalo $[a,b]$ de integração
* O maior grau $P$ da função $f(x)$
* O tipo de Quadratura (GL ou GLL)
* A base a ser usada a depender dos parâmetros $\alpha$ e $\beta$ (no nossos exemplos vamos considerar a base de Legendre somente)

```python
from typing import Callable

def integrar_dg(f: Callable[[float], float],  a: float, b: float, P: int,  quad_type: str = 'GL',
                alpha: float = 0.0, beta: float = 0.0) -> float:
    """Integra a função f(x) no domínio [a, b] utilizando nossas rotinas generalizadas."""
    # Fabricamos as matrizes de zeros e pesos dinamicamente (Legendre: alpha=0, beta=0)
    xi, wi = jacobi_gauss_quad(P, alpha, beta, quad_type)

    m = (b - a) / 2.0
    c = (b + a) / 2.0

    integral = 0.0
    for i in range(len(xi)):
        integral += wi[i] * f(m * xi[i] + c)

    return m * integral

```

### Exemplo 1
Integrar $f(x) = x^{10}$ no intervalo $[0, 1]$ com quadratura de Gauss-Jacobi-Legendre


```python
# =====================================================================
# EXEMPLO 1: f(x) = x^10 no intervalo [0, 1]
# =====================================================================
# Integral Analítica = 1/11 (aprox 0.090909...)

f_10 = lambda x: x**10
res_10_gl = integrar_dg(f_10, 0.0, 1.0, P = 10, quad_type='GL')
print(f"Exemplo 1: Integral de x^10 em [0,1]")
print(f"Analítico:  {1/11:.7f}")
print(f"Numérico:   {res_10_gl:.7f}")
```

Na **aula 00**, vimos que cada elemento da **Matriz de Massa** é definida, conforme artigo do Moura, por

$$\mathcal{M}_{mn} = \int_{\Omega_{pd}} \phi_m \phi_n d \xi$$

Da **aula 01**, se as funções de aproximação são **Polinômios Ortogonais**, o resultado dessa integral será definido pela delta de Kronecker $\delta_{mn}$.

No caso deles serem da **Polinômios de Legendre**, o resultado será a $\delta_{mn}$ vezes uma ponderação, isto é:

$$\mathcal{M}_{mn} = \int_{-1}^1 P^{0,0}_m P^{0,0}_n d \xi = \dfrac{2}{2n+1}\delta_{mn}$$

ou ainda
$$\mathcal{M}_{mn} =
\begin{cases}
0 & \ m \neq n \\
\dfrac{2}{2n+1} & \ m = n
\end{cases}
$$

---
Verifique que tal propriedade é válida integrando no intervalo padrão $[-1, 1]$ as combinações:  
* $P_2^{0,0} \cdot P_2^{0,0}$
* $P_2^{0,0}\cdot P_3^{0,0}$

Leve em conta que a multiplicação de dois polinômios gera um outro polinômio de grau $m+n$

---
**Curiosidade:** [Ortogonalidade para Polinomios de Jacobi](https://en.wikipedia.org/wiki/Jacobi_polynomials#Orthogonality)

```python
# =====================================================================
# EXEMPLO 6: Matriz de Massa (Ortogonalidade com Polinomios de Legendre)
# =====================================================================
# Integrando P_2(x) * P_2(x) no domínio padrão [-1, 1].
# Funções iguais devem ter integral = 2/(2n+1).
f_massa22 = lambda x: JacobiP(x, 2, 0.0, 0.0) * JacobiP(x, 2, 0.0, 0.0)
res_massa22 = integrar_dg(f_massa22, -1.0, 1.0, P=4, quad_type='GL')
print("Exemplo 6a: Elemento M22 da Matriz de Massa : (P_2 * P_2) ")
print(f"Analítico:  0.4000000")
print(f"GL (P=4):   {res_massa22:.7f}\n")

# Integrando P_2(x) * P_3(x) no domínio padrão [-1, 1].
# Funções ortogonais devem ter integral = 0.
f_massa23 = lambda x: JacobiP(x, 2, 0.0, 0.0) * JacobiP(x, 3, 0.0, 0.0)
res_massa23 = integrar_dg(f_massa23, -1.0, 1.0, P=5, quad_type='GL')
print("Exemplo 6b: Elemento M23 da Matriz de Massa : (P_2 * P_3) ")
print(f"Analítico:  0.0000000 (Pela Ortogonalidade)")
print(f"GL (P=5):   {res_massa23:.7f} <-- Os zeros fora da diagonal da Matriz M!")
```

```python
# =====================================================================
# EXEMPLO 7: Matriz de Rigidez
# =====================================================================
# Integrando P_1(x) * P'_2(x) no domínio padrão [-1, 1].
#    P_1(x) = x.      P'_2(x) = 3x.       Produto = 3x^2.
# =====================================================================
# Integral analítica de 3x^2 em [-1, 1] = x^3|_{-1}^{1} = 1 - (-1) = 2.0
# Grau = 1 + 1 = 2. GL com N=2 já é suficiente.
f_rigidez = lambda x: JacobiP(x, 1, 0.0, 0.0) * DJacobiP(x, 2, 0.0, 0.0)
res_rigidez = integrar_dg(f_rigidez, -1.0, 1.0, P=2, quad_type='GL')

print("Exemplo 7: Matriz de Rigidez (P_1 * P'_2)")
print(f"Analítico:  2.0000000")
print(f"GL (P=2):   {res_rigidez:.7f} <-- Integração exata da derivada!")
```

### **Decomposição Polinomial**

Considere um polinômio
$$f(x)\in\mathcal{P}_{2Q-k}([-1,1]).$$

Deseja-se mostrar que ele pode ser escrito na forma
$$f(x)=\sum_{i=0}^{Q-1}f(x_i)h_i(x)+s(x)r(x),$$

em que
$$
\begin{aligned}
h_i(x_j)&=\delta_{ij}, & h_i(x)&\in\mathcal{P}_{Q-1},\\
s(x_i)&=0,             & s(x)&\in\mathcal{P}_{Q},\\
&& r(x)&\in\mathcal{P}_{Q-k}.
\end{aligned}
$$

Essa decomposição decorre diretamente da teoria de interpolação polinomial.

---
Primeiramente, define-se o polinômio interpolador de Lagrange
$$I_{Q-1}f(x)=\sum_{i=0}^{Q-1}f(x_i)h_i(x),$$

o qual pertence ao espaço
$$I_{Q-1}f(x)\in\mathcal{P}_{Q-1},$$

uma vez que cada função de base $h_i(x)$ possui grau no máximo $Q-1$.

Em seguida, define-se o erro de interpolação
$$e(x)=f(x)-I_{Q-1}f(x).$$

Como o interpolador coincide com $f(x)$ em todos os nós de interpolação,
$$I_{Q-1}f(x_i)=f(x_i),$$

segue que

$$e(x_i)=0,\qquad i=0,\ldots,Q-1.$$

Portanto, $e(x)$ possui os $Q$ nós de interpolação como raízes. Pela propriedade fundamental da divisibilidade de polinômios, todo polinômio que possui essas raízes é divisível por
$$ s(x)=\prod_{i=0}^{Q-1}(x-x_i), $$

de modo que existe um polinômio $r(x)$ tal que
$$ e(x)=s(x)r(x). $$

Substituindo a definição de $e(x)$,
$$f(x)-I_{Q-1}f(x)=s(x)r(x),$$

ou, equivalentemente,
$$\boxed{f(x)=\sum_{i=0}^{Q-1}f(x_i)h_i(x)+s(x)r(x).}$$

Resta determinar o grau de $r(x)$. Como
$$\deg(f)\le 2Q-k$$

e
$$\deg(I_{Q-1}f)\le Q-1,$$

tem-se
$$\deg(e)\le 2Q-k.$$

Além disso,
$$\deg(s)=Q,$$

logo,
$$\deg(r)=\deg(e)-\deg(s)\le(2Q-k)-Q=Q-k.$$

Assim,
$$r(x)\in\mathcal{P}_{Q-k},$$

concluindo a demonstração.

---

### **O uso do Polinômio** $P_{Q-2}^{\alpha+1,\beta+1}$ **no GLL**

A quadratura de Gauss-Jacobi procura uma aproximação da forma
$$ \int_{-1}^{1}f(x)W(x),dx \approx \sum_{i=1}^{N}w_if(x_i), $$
onde $ W(x)=(1-x)^\alpha(1+x)^\beta.$

Os nós $(x_i)$ são escolhidos para que a regra seja exata para polinômios do maior grau possível. Um resultado clássico da teoria de ortogonalidade mostra que isso acontece **se, e somente se, os nós forem as raízes do polinômio ortogonal de grau (N)** em relação ao peso $(w(x))$.

Ou seja,
$$ x_i \quad\Longleftrightarrow\quad P_N^{\alpha,\beta}(x_i)=0.$$
Até aqui não existe nenhuma restrição nos extremos.

---

**Agora imponha que $\pm 1$ pertençam à malha**  
Na quadratura de Lobatto dizemos:  
> ''Independentemente do que seja ótimo, eu quero que os pontos $-1$ e $+1$ estejam presentes.''

Então dois dos nós deixam de ser incógnitas. Restam apenas os pontos internos para serem escolhidos. Isto é se existam $(n = N)$ nós.

Na Gauss-Jacobi temos $(n = N)$ incógnitas. Na Gauss-Lobatto temos **dois nós** já conhecidos; e, portanto, apenas $(n = N-2)$ incógnitas. Ou seja, o problema de otimização mudou.

---

**Como escrever um polinômio que obrigatoriamente possua essas raízes?**  
Se sabemos que
$$x=-1,\qquad x=1$$
devem ser raízes, então qualquer polinômio admissível pode ser escrito como
$$ \pi(x)=(1-x)(1+x)q(x). $$
Toda a liberdade restante está em $(q(x))$. Até aqui não apareceu Jacobi nenhum.

---

**Agora entra a ortogonalidade**  
Na quadratura de Gauss, o **polinômio ótimo satisfaz** $R(f)=0$
$$R(f) = \int_{-1}^{1} W(x) P_N^{\alpha,\beta}(x)r(x)dx =0 $$
para qualquer polinômio $(r(x))$ de grau menor. Na Lobatto, como o polinômio admissível é
$$\pi(x)=(1-x)(1+x)q(x),$$

a condição de ortogonalidade passa a ser
$$\int_{-1}^{1}W(x)(1-x)(1+x)q(x)r(x)dx=0.$$

Agora basta juntar os pesos. Como
$$W(x)=(1-x)^\alpha(1+x)^\beta,$$

temos
$$(1-x)(1+x)W(x) = (1-x)^{\alpha+1}(1+x)^{\beta+1}.$$

Ou seja, o novo problema de ortogonalidade é
$$\boxed{\int_{-1}^{1}q(x)r(x)(1-x)^{\alpha+1}(1+x)^{\beta+1}dx=0.}$$

---

**E quem é o polinômio ortogonal desse novo peso?**

Mas essa é exatamente a definição dos polinômios de Jacobi! Só que agora o peso não é mais
$$(1-x)^\alpha(1+x)^\beta,$$
e sim
$$(1-x)^{\alpha+1}(1+x)^{\beta+1}.$$

Logo, o polinômio ortogonal passa a ser
$$P_n^{\alpha+1,\beta+1}(x).$$

Portanto,

$$q(x)=P_{N-2}^{\alpha+1,\beta+1}(x).$$

Finalmente,
$$\boxed{\pi(x)=(1-x)(1+x)P_{N-2}^{\alpha+1,\beta+1}(x).}$$

Observe que o deslocamento dos parâmetros **não foi imposto**. Ele apareceu porque o fator
$$(1-x)(1+x)$$

foi incorporado ao peso da ortogonalidade.

---




#### Uma interpretação física (ou geométrica)

Existe uma maneira bastante intuitiva de enxergar isso. Os polinômios de Jacobi são "ótimos" para um determinado peso. Quando você obriga o polinômio a ser nulo nos extremos, está retirando dois graus de liberdade.

Esses dois fatores
$$(1-x)(1+x)$$
passam a fazer parte da estrutura fixa do polinômio Consequentemente, o problema restante é escolher apenas a parte interna $q(x)$.

Só que essa parte interna "vive" agora em um espaço de funções cujo peso ficou multiplicado por
$$(1-x)(1+x).$$

Ou seja, o problema de otimização restante acontece naturalmente no espaço
$$L^2\left([-1,1],(1-x)^{\alpha+1}(1+x)^{\beta+1}\right),$$
e os polinômios ortogonais desse espaço são exatamente
$$P_n^{\alpha+1,\beta+1}(x).$$

---

Esse raciocínio também explica por que, no caso de **Legendre** $(\alpha=\beta=0)$, aparece o conhecido resultado (**forma apresentada no site do Wolfram!!**)
$$(1-x^2)P_{N-2}^{1,1}(x) \propto (1-x^2)P_{N-1}'(x),$$

que é uma consequência da identidade
$$
\frac{d}{dx}P_n^{\alpha,\beta}(x) =\frac{n+\alpha+\beta+1}{2}
P_{n-1}^{\alpha+1,\beta+1}(x).
$$



Assim, os nós internos de Gauss–Lobatto–Legendre podem ser vistos tanto como as raízes de $P_{N-2}^{1,1}$ quanto, de forma equivalente, como as raízes da derivada de $P_{N-1}$.
